```python
yeni_kolonlar = {
    'PO_ID': 'Siparis_ID',
    'Supplier': 'Tedarikci',
    'Order_Date': 'Siparis_Tarihi',
    'Delivery_Date': 'Teslimat_Tarihi',
    'Item_Category': 'Urun_Kategorisi',
    'Order_Status': 'Siparis_Durumu',
    'Quantity': 'Miktar',
    'Unit_Price': 'Birim_Fiyat',
    'Negotiated_Price': 'Anlasilan_Fiyat',
    'Defective_Units': 'Kusurlu_Urun_Sayisi',
    'Compliance': 'Uyum_Skoru'
}

# DataFrame kolonlarını güncelleme
df.rename(columns=yeni_kolonlar, inplace=True)
```

# Tablonun kategorik değişkenlerine ait ingilizce kayıtları TÜRKÇE yapan kod

```python
# Ürün Kategorileri Değişimi
kategori_map = {
    'Office Supplies': 'Ofis_Malzemeleri',
    'MRO': 'MRO_Bakim_Onarim',
    'Packaging': 'Ambalaj_Paketleme',
    'Raw Materials': 'Hammadde',
    'Electronics': 'Elektronik'
}
df['Urun_Kategorisi'] = df['Urun_Kategorisi'].map(kategori_map)

# Sipariş Durumu Değişimi
durum_map = {
    'Cancelled': 'Iptal_Edildi',
    'Delivered': 'Teslim_Edildi',
    'Partially Delivered': 'Kismen_Teslim_Edildi',
    'Pending': 'Beklemede'
}
df['Siparis_Durumu'] = df['Siparis_Durumu'].map(durum_map)

# Uyum Skoru Değişimi
uyum_map = {
    'Yes': 'Uygun', # veya 'Evet'
    'No': 'Uygun_Degil' # veya 'Hayir'
}
df['Uyum_Skoru'] = df['Uyum_Skoru'].map(uyum_map)
```
